```python
notas = [22, 34, 15, 26, 18, 22] #slicing

sub_notas = notas[1:-2]  # no coje el último indice - el anterior
print(sub_notas)
notas[1:] == notas[1:6] == notas[1:len(notas)]
print("----", notas[1:] == notas[1:6] == notas[1:len(notas)])
print(notas[1:])
print(notas[1:6])
print(notas[1:len(notas)])
```

## Modificar 

- listas mutables

```python
meses = ["enero", "febrero", "marzo"]

meses[1] = "Abril"
print(meses)
```

## Añadir elementos

- `append, insert`

```python
meses = ["enero", "febrero", "marzo"]

meses.append("mayo")
print(meses)
```

```python
meses.insert(3, "abril")
print(meses)
```

```python
lst = []

for mes in meses:
    if 'a' in mes:
        lst.append(mes)

print(lst)
```

## Eliminar elementos

- `remove` por valor la 1a occurencia; si indicamos un elemento que no está en la lista devuelve error
- `pop` por indice Y permite recuperarlo

```python
meses.remove('marzo')
print(meses)
```

```python
eliminado = meses.pop(1)
print('eliminado', eliminado)
print('meses',meses)
```
